# Tensorflow tutorial 진행하기

- Keras 를 사용해서 데이터세트를 로드
  - MNIST 데이터셋 로드
- 이미지 분류 신경망 머신러닝 모델 빌드
- 신경망 훈련 진행
- 모델 정확도 평가


```python
import tensorflow as tf

print("tensorflow version:" , tf.__version__)

# colab 외에 환경에서 진행하는 경우, pip로 tensorflow를 설치 및 업데이트가 되었는지 체크
```

```python
# 데이터 세트 로드하기

mnist = tf.keras.datasets.mnist

(x_train, y_train), (x_test, y_test) = mnist.load_data()

# 데이터를 정수에서 부동소수점 타입으로 변환
x_train, x_test = x_train / 255.0, x_test / 255.0
```

```python
# 신경망 머신러닝 모델 빌드

model = tf.keras.models.Sequential([
    tf.keras.layers.Flatten(input_shape=(28,28)),
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dropout(0.2),
    tf.keras.layers.Dense(10, activation="softmax")
    # softmax 는 예측값을 각 범주별 확률분포로 변환해주는 함수
])

# 모델을 컴파일해서 학습 준비 시키기
model.compile(optimizer="adam", loss="sparse_categorical_crossentropy", metrics=['accuracy'])
# sparse_categorical_crossentropy 는 분류 문제 / 분류 레이블이 원핫인코딩 레이블 대신 정수레이블일때 주로 사용하는 손실함수
```

```python
# 빌드한 모델 예측 시키기
  # 맨 처음 불러온 모델은 학습이 전혀 진행되지 않은 상태

predictions = model(x_train[:1]).numpy() # 데이터 한개 입력
# model() 또는 model.predict()는 새로운 입력 데이터에 대한 예측을 수행하도록 지시할때 사용 -> 모델의 출력 레이어(=마지막 레이어)에 따라 예측된 값을 반환
# model.fit() 은 모델을 훈련 데이터에 맞춰지도록 학습시키는데 사용, 이 과정에서 가중치를 조정하고 손실함수를 최소화하고 성능지표를 최적화함.
  # 훈련 특징데이터(x_train)와 정답레이블(y_train)과, 옵션으로 검증데이터(x_val, y_val), epoch수, batch크기, 출력방법지정 등의 매개변수를 입력받고, 훈련진행상황 로그를 포함하는 History 객체를 반환.
print(predictions)
print(tf.nn.softmax(predictions).numpy())
```

```python
# 훈련용 손실함수를 정의하기

loss_fn = tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True)
# 손실값은 모델이 올바르게 예측하는 경우 0이 되고, 맨초기 모델은 무작위에 가까운 확률을 제공하기 때문에 2.3 정도에 근접한다.

loss_fn(y_train[:1], predictions).numpy()
```

```python
# 새로 만든 손실함수로 모델 다시 컴파일해서 학습 준비 시키기

model.compile(optimizer="adam", loss=loss_fn, metrics=['accuracy'])
```

```python
# 모델 학습 시키기

model.fit(x_train, y_train, epochs = 5)
```

```python
# 데이터셋에 적합하게 학습된 모델 평가

model.evaluate(x_test, y_test, verbose=2)
```

```python
# 모델이 확률을 반환하도록 하려면 훈련된 모델을 래핑해서 소프트맥스를 첨부

probability_model = tf.keras.Sequential([
    model,
    tf.keras.layers.Softmax()
])

probability_model(x_test[:5])
```
